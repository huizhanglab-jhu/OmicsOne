```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Example data for the heatmap
data = {
    "Gene1": [1.2, 0.8, 1.5],
    "Gene2": [0.5, 1.0, 0.9],
    "Gene3": [1.1, 1.3, 0.7]
}
index_labels = ["Sample1", "Sample2", "Sample3"]

# Create a DataFrame
df = pd.DataFrame(data, index=index_labels)

# Create the heatmap
plt.figure(figsize=(8, 6))  # Set figure size
sns.heatmap(df, annot=True, fmt=".2f", cmap="coolwarm", cbar=False)

# Customize the plot
plt.title("Heatmap Example")
plt.xlabel("Genes")
plt.ylabel("Samples")

# Show the plot
plt.tight_layout()
plt.show()
```
