```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Create a figure with a specific size
fig = plt.figure(figsize=(10, 6))

# Create two axes with specific height ratios
# The first axis (ax1) will be larger for the bar plot
# The second axis (ax2) will be smaller for the heatmap
gs = fig.add_gridspec(2, 1, height_ratios=[4, 1], hspace=0.02)
ax1 = fig.add_subplot(gs[0])
ax2 = fig.add_subplot(gs[1])

# Example data (replace with your actual data)
# For bar plot
bar_data = pd.DataFrame({
    'Sample': range(50),
    'Correlation': np.random.uniform(0.3, 0.5, 50)
}).sort_values('Correlation', ascending=False)

# For heatmap
heatmap_data = pd.DataFrame(
    np.random.choice(['Status1', 'Status2', 'Status3'], size=(2, 50)),
    index=['Survival', 'PlatinumStatus'],
    columns=bar_data.index
)

# Create bar plot
sns.barplot(data=bar_data, x='Sample', y='Correlation', 
            color='blue', ax=ax1)

# Remove x-axis labels from bar plot
ax1.set_xticklabels([])
ax1.set_xlabel('')

# Customize bar plot
ax1.set_title('Sample-wise mRNA-protein correlation')
ax1.set_ylabel('Index (Spearman correlation)')

# Create heatmap
# Create custom colormap if needed
# Create heatmap
# Create numerical mapping for categorical data
category_map = {'Status1': 0, 'Status2': 1, 'Status3': 2}
heatmap_data_numeric = heatmap_data.replace(category_map)

# Define custom colors for the heatmap
colors = ['green', 'pink', 'gray']
custom_cmap = sns.color_palette(colors)

# Create heatmap with numeric data
# sns.heatmap(heatmap_data_numeric, cmap=custom_cmap, cbar=False, ax=ax2)
# Create heatmap with numeric data and show colorbar
sns.heatmap(heatmap_data_numeric, cmap=custom_cmap, cbar=True, ax=ax2,
            cbar_kws={'ticks': [0.5, 1.5, 2.5],  # Center of each category
                     'label': '',  # Remove default label
                     'orientation': 'horizontal'})  # Make colorbar horizontal

# Modify colorbar ticks to show category names
colorbar = ax2.collections[0].colorbar
colorbar.set_ticklabels(['Status1', 'Status2', 'Status3'])


# Remove x-axis labels from heatmap if desired
ax2.set_xticklabels([])
ax2.set_xlabel('Sample')

# Adjust layout
plt.tight_layout()
plt.show()
```
