```python
path = r"E:\project\LijunChen\PDX\PDX_batch1-7_proteomics\header_expression_matrix.tsv"
```

```python
import pandas as pd
df = pd.read_csv(path, sep="\t", index_col=0)
```

```python
df
```

```python
rows = []
import re
for index,row in df.iterrows():
    m = re.search(r'Batch[2-7]', index)
    b = m.group(0) if m else 'Batch1'
    for col in df.columns:
        i = str(row[col]).strip()
        rows.append((i, col, b))
        
df = pd.DataFrame(rows, columns=['case_id', 'channel', 'batch'])
```
