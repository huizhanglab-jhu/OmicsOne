```python
data_dir = r"E:\project\chitra"
from pathlib import Path
out_dir = Path(data_dir) / "output"

out_dir.mkdir(exist_ok=True, parents=True)
```

```python
from pathlib import Path

data_path = Path(data_dir) / "Chitra_global_Report.tsv.txt"

import pandas as pd
data = pd.read_csv(data_path, sep="\t")
```

```python
data
```

```python
data['Protein'] = [i.split(";")[0] for i in data['PG.ProteinAccessions']]
data.set_index('Protein', inplace=True)
data
```

```python
rows = []

for index,row in data.iterrows():
    proteins = row['PG.ProteinAccessions'].split(";")
    if pd.isna(row['PG.Genes']):
        continue
    genes  = row['PG.Genes'].split(";")

    # print(proteins, genes)
    for protein, gene in zip(proteins, genes):
        rows.append([protein, gene])


with open(out_dir / "protein_gene_map.tsv", "w") as f:
    f.write("Protein\tGene\n")
    for protein, gene in rows:
        f.write(f"{protein}\t{gene}\n")
```

```python
data.drop(columns=['PG.ProteinAccessions','PG.Genes','PG.Organisms','PG.ProteinDescriptions','PG.ProteinNames'], inplace=True)
```
