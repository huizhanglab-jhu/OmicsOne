```python
import os,re,sys
import pandas as pd
import numpy as np
import subprocess
```

```python
import shutil
```

```python
# panoply_cons_cluster:
#   clustering_sd_threshold: 2   # threshold for filtering data before consensus clustering
#   clustering_na_threshold: 0.5 # 

clustering_sd_threshold=2
clustering_na_threshold=0.5
```

```python
analysis_dir = r"/Users/yingweihu/project/panoply/tutorial-brca-input"
cluster_dir = os.path.join(analysis_dir,"cluster")
if not os.path.exists(cluster_dir):
    os.mkdir(cluster_dir)
code_dir = r"/Users/yingweihu/Documents/GitHub/PANOPLY/src/panoply_cons_clust"
for file_name in os.listdir(code_dir):
    src_file = os.path.join(code_dir,file_name)
    dst_file = os.path.join(cluster_dir,file_name)
    if os.path.isfile(src_file):
        shutil.copy(src_file, dst_file)
        print(f"copy {src_file} to {dst_file}")
r_script = os.path.join(code_dir, "panoply_kmeans_consensus.R")
filter_dir = os.path.join(analysis_dir,"filter")
if not os.path.exists(filter_dir):
    os.mkdir(filter_dir)

temp_dir = os.path.join(analysis_dir,"temp")
if not os.path.exists(temp_dir):
    os.mkdir(temp_dir)


```

```python


command = [
    "Rscript",
    r_script,
    "-i", analysis_dir,
    "-u", "2",
    "-v", "3",
    "-b", "1000",
    "-s", clustering_sd_threshold,
    "-l", 'con_clust',
    "-t", "yhu39",
    "-f", filter_dir,
    "-c", cluster_dir,
    "-d", temp_dir,
    "-z", code_dir
]
```

```python
# Run the subprocess
try:
    result = subprocess.run(
        command,
        capture_output=True,  # Capture the standard output and error
        text=True,            # Decode output as text
        check=True            # Raise an error if the command fails
    )
    # Print standard output
    print("R Script Output:")
    print(result.stdout)
except subprocess.CalledProcessError as e:
    # Handle errors
    print("Error occurred:")
    print(e.stderr)
```
