```python
df1_sub = df_peptide_2025.iloc[:, 2:]
df2_sub = df_peptide_2024.iloc[:, 2:]

# 确保列数一致
if df1_sub.shape != df2_sub.shape:
    result = "samples_different"
else:
    # 创建布尔掩码，记录两者是否每个元素都相等（考虑误差）
    comparison = []

    for col in df1_sub.columns:
        s1 = df1_sub[col]
        s2 = df2_sub[col]

        # 两列是否都是数字
        if pd.api.types.is_numeric_dtype(s1) and pd.api.types.is_numeric_dtype(s2):
            # 使用 np.isclose 判断误差小于 1e-6，且同时处理 NaN 比较
            comp = np.isclose(s1, s2, atol=1e-6, equal_nan=True)
        else:
            # 对于非数字型，直接比较是否一致（含 None/NaN）
            comp = (s1.fillna("NA") == s2.fillna("NA"))

        comparison.append(comp)

    # 所有比较都为 True 才返回“是”
    result = "samples_same" if np.all(np.column_stack(comparison)) else "samples_different"

print(result)
```

```python
def compare_dataframes(df1, df2):
    # 检查行数和列数是否相同
    shape,sequence,sampleValues = True,True,True
    if df1.shape != df2.shape:
        shape = False
        print("shape_different")

    # 检查sequence是否相同
    if df1.iloc[:,0].equals(df2.iloc[:,0]):
        print ("sequence_same")
    else:
        sequence = False
        print("seqence_different")

    # 检查每个元素是否相等（考虑误差）
    df1_sub = df1.iloc[:, 2:]
    df2_sub = df2.iloc[:, 2:]
    comparison = []

    for col in df1_sub.columns:
        s1 = df1_sub[col]
        s2 = df2_sub[col]

        # 两列是否都是数字
        if pd.api.types.is_numeric_dtype(s1) and pd.api.types.is_numeric_dtype(s2):
            # 使用 np.isclose 判断误差小于 1e-6，且同时处理 NaN 比较
            comp = np.isclose(s1, s2, atol=1e-6, equal_nan=True)
        else:
            # 对于非数字型，直接比较是否一致（含 None/NaN）
            comp = (s1.fillna("NA") == s2.fillna("NA"))

        comparison.append(comp)

    # 所有比较都为 True 才返回“是”
    if np.all(np.column_stack(comparison)):
        sampleValues = True
        print("SamplesValues same")
    else:
        sampleValues = False
        print("SamplesValues different")
    
    
    if shape and sequence and sampleValues:
        print("dataframes_same")
        return "dataframes_same"
```
